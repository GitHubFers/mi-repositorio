4. Utiliza histogramas para determinar la distribución de los valores representados en cada variable.
*   ¿Se corresponde con lo obtenido en el cálculo de asimetría? Como verás, los datos reales son más complejos que la teoría. Por esta razón, recuerda siempre acompañar el análisis de la asimetría con algún gráfico como un histograma.

**NOTA**. Para esto también puedes ocupar los gráficos `kde` ([kernel density estimation](https://www.cienciadedatos.net/documentos/pystats02-kernel-density-estimation-kde-python.html)) que crean una curva continua y suave expandiendo la idea del histograma.

```python
#Histograma + curva KDE por cada variable numerica
num_cols = df.select_dtypes(include='number').columns
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
for ax, col in zip(axes.flat, num_cols):
    sns.histplot(df[col], kde=True, ax=ax, bins=40)
    ax.set_title(f'{col} (asimetría = {df[col].skew():.2f})')
for ax in axes.flat[len(num_cols):]:
    ax.set_visible(False)   #oculta los ejes sobrantes
plt.tight_layout()
plt.show()
```

image.png

<font color="orange">**Justificación:** Sí, en general los histogramas **coinciden** con la asimetría calculada:
* `person_age`, `person_emp_length`, `cb_person_cred_hist_length`, `loan_amnt` y `loan_percent_income` tienen la masa de datos a la izquierda y cola larga a la derecha (asimetría positiva).
* `person_income` es la más extrema (asimetría ≈ 32.9): casi todo el histograma se concentra en un solo bloque porque hay ingresos de hasta 6,000,000; **los valores atípicos aplastan el gráfico** y hace falta acotar el eje para verlo bien.
* `loan_int_rate` es la más simétrica (≈ 0.21), aunque no es una campana perfecta: se ve algo multimodal/plana.
* `loan_status` es binaria (0/1), por lo que su asimetría (1.36) solo refleja que hay más ceros (≈ 78 %) que unos (≈ 22 %) y el histograma son dos barras.
Esto confirma que conviene acompañar la asimetría con un gráfico: los valores extremos (como edades de 144 años o antigüedad laboral de 123 años) distorsionan tanto el estadístico como la gráfica.</font>

*   Como podrás observar hay valores atípicos en todas las variables. Ejecuta el siguiente código para identificar los valores atípicos en la variable `person_age`

**Interpretación Resultados**

<font color="orange">En los boxplots casi todas las variables tienen puntos fuera de los bigotes (valores atípicos), sobre todo hacia arriba. En `person_age` el método del rango intercuartílico (IQR) usa Q1 = 23 y Q3 = 30, por lo que IQR = 7. Los límites son: inferior = 23 − 1.5·7 = **12.5** y superior = 30 + 1.5·7 = **40.5**. Se detectan **1,494 registros atípicos** (≈ 4.6 % de los datos), todos por encima del límite superior, con edades entre 41 y 144 años. La mayoría son personas mayores de 40 años, que son valores posibles pero poco frecuentes en este conjunto (población mayormente joven, mediana de 26). Sin embargo, hay casos **claramente erróneos** (5 registros con más de 100 años, hasta 144), que deberían corregirse o eliminarse antes de modelar. Los valores 41-80 años pueden ser legítimos y no conviene borrarlos automáticamente.</font>

```python
#COMENTE CADA LINEA DE CODIGO EXPLICANDO SU USO

#Calcula el percentil 25 (primer cuartil, Q1) de la edad
percentile_25 = df["person_age"].quantile(0.25)
#Calcula el percentil 75 (tercer cuartil, Q3) de la edad
percentile_75 = df["person_age"].quantile(0.75)

#Rango intercuartilico (IQR): distancia entre Q3 y Q1, es decir, donde esta el 50% central de los datos
iqr = percentile_75 - percentile_25

#Limite superior: todo lo que pase de Q3 + 1.5*IQR se considera atipico
upper_limit = percentile_75 + 1.5 * iqr
#Limite inferior: todo lo que quede debajo de Q1 - 1.5*IQR se considera atipico
lower_limit = percentile_25 - 1.5 * iqr

#Filtra el dataframe: se queda con las filas cuya edad esta por debajo del limite inferior (<) o (|) por encima del superior (>)
IQR_outliers = df[(df["person_age"] < lower_limit) | (df["person_age"] > upper_limit)]
#Muestra los registros atipicos
IQR_outliers
```
